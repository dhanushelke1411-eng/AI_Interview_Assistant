```python
from transformers import pipeline

interviewer = pipeline(
    "text-generation",
    model="gpt2"
)

print("AI Interview Assistant Ready")
```

```python
job_role = input("Enter Job Role: ")

prompt = f"Generate 5 interview questions for a {job_role} position."

response = interviewer(
    prompt,
    max_length=120,
    num_return_sequences=1
)

print("\nInterview Questions:\n")
print(response[0]['generated_text'])
```

```python
job_role = input("Enter Job Role: ")

prompt = f"""
You are an AI Interview Coach.

Generate:
1. 5 interview questions
2. Short sample answers
3. Interview tips

for a {job_role} position.
"""

response = interviewer(
    prompt,
    max_length=300,
    num_return_sequences=1
)

print("\nAI Interview Preparation:\n")
print(response[0]['generated_text'])
```
